```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
```

```python
#data loading
df=pd.read_csv('iris.csv')
df.head()
```

```python
print("Dataset info:", df.info())
print("Shape of the dataset:", df.shape)
print("Statistical Summary:", df.describe())
```

```python
print("Species Distribution:", df["Species"].value_counts())
```

```python
#checking missing values
df.isnull().sum()
```

```python
#checking for duplicated rows
print("Number of duplicated rows:", df.duplicated().sum())
```

```python
label_encoder = LabelEncoder()
df['Species_Encoded']=label_encoder.fit_transform(df['Species'])
```

```python
print("Mapping of encoded species:")
for i, class_name in enumerate(label_encoder.classes_):
  print(f"{class_name} ---> {i}")

print("\nFirst 5 rows after encoding:")
print(df.head())
```

```python
#Feature Scaling
feature_cols=[
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]
X=df[feature_cols]

scaler_standard = StandardScaler()
X_scaled_standard=scaler_standard.fit_transform(X)

minimax_scaler=MinMaxScaler()
X_scaled_minimax=minimax_scaler.fit_transform(X)
```

```python
df_standard = pd.DataFrame(X_scaled_standard, columns=feature_cols)
print("\nFirst 5 rows after Standardization (Mean=0, SD=1):")
print(df_standard.head())
```
